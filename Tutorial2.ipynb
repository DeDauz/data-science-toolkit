```python
#1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])


print("this is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4
random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3.Create a numpy array of ones and a numpy array of zeros

ones_array  = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones:")
print(ones_array)

print("Array of Zeros:")
print(zeros_array)
```

```python
#4.Create a random numpy array of size 6x3 and transpose it

array_6x3 = np.random.rand(6,3)
print("Random array 6x3:")
print(array_6x3)

transposearray = np.transpose(array_6x3)
print("Transposed array:")
print(transposearray)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product

array3x4 = np.random.rand(3,4)
print("Random 3x4 array:")
print(array3x4)
array4x5 = np.random.rand(4,5)
print("Random 4x5 array:")
print(array4x5)

dotproduct = np.dot(array3x4,array4x5)
print("Dot product of two array:")
print(dotproduct)
```
